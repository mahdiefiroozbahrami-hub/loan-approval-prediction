Filling Missing value of LoanAmount column

```python
loan_df['LoanAmount'].describe()
```

First, we examine the mean, standard deviation, and quantiles to check whether the distribution is normal or skewed.

```python
l_a_median=loan_df['LoanAmount'].median()
l_a_mean=loan_df['LoanAmount'].mean()
print(f"Mean: {l_a_mean} , Median {l_a_median}")
```

We observe that the mean of LoanAmount is about 146 and the median is 128, indicating that the data is right-skewed. Therefore, we first impute the missing values with the median (since the mean is sensitive to outliers), and then remove outliers using the IQR method.

```python
loan_df['LoanAmount'].isna().sum()
```

```python
loan_df['LoanAmount'].fillna(loan_df['LoanAmount'].median(), inplace=True)
```

```python
loan_df['LoanAmount'].isna().sum()
```

There is no null values in LoanAmount column anymore.

```python
loan_df['LoanAmount'].describe()
```

The LoanAmount distribution is right-skewed, with the minimum and maximum values being considerably distant from the mean, indicating the presence of outliers.

```python
q1=loan_df['LoanAmount'].quantile(.25)
q3=loan_df['LoanAmount'].quantile(.75)
IQR=q3-q1
lower=q1-1.5*IQR
upper=q3+1.5*IQR
loan_df_cleaned=loan_df[(loan_df['LoanAmount']>=lower) & (loan_df['LoanAmount']<=upper)]
```

Outliers were detected using the interquartile range (IQR) method, and subsequently omitted by retaining only the values within the calculated lower and upper bounds.

A new feature, TotalApplicantIncome, was created by summing the income of the applicant (ApplicantIncome) and the co-applicant (CoapplicantIncome). This feature represents the total household income available for loan repayment and is expected to have a stronger relationship with the target variable (Loan_Status) compared to individual incomes.

```python
loan_df_reduced['TotalApplicantIncome'].describe()
```

The TotalApplicantIncome feature, created by summing the ApplicantIncome and CoapplicantIncome, is right-skewed according to loan_df['TotalApplicantIncome'].describe(). Therefore, it should be transformed using a logarithmic scale to reduce skewness and mitigate the effect of extreme values

```python
loan_df_reduced.loc[:,'Log_TotalApplicantIncome'] = np.log1p(loan_df_reduced['TotalApplicantIncome'])
```

```python
loan_df_reduced[['Log_TotalApplicantIncome', 'Loan_Status_num']].corr()
```

The correlation shows that there is a week relation between Log_TotalApplicantIncome and Loan_Status_num.Here I want to test both modes(with and without Log_TotalApplicantIncome)


```python
loan_df_reduced.head()
```

```python
loan_df_reduced['Log_TotalApplicantIncome'].describe()
```

After applying the logarithmic transformation to TotalApplicantIncome, the mean and median values are close, indicating a more balanced distribution. The standard deviation is reduced, and the effect of extreme values is mitigated, making this feature more suitable for modeling with Logistic Regression, KNN, and ANN

Another new feature 

```python
loan_df_reduced.loc[:,'Log_Loan_Income_Ratio']=loan_df_reduced['Log_LoanAmount']/loan_df_reduced['Log_TotalApplicantIncome']
```

A new feature, Log_Loan_Income_Ratio, was created by dividing the logarithmically transformed LoanAmount (Log_LoanAmount) by the logarithmically transformed TotalApplicantIncome (Log_TotalApplicantIncome).
